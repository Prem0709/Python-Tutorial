`1. df['Phone_Number'].fillna('')`
- Handles missing values (NaN) by replacing them with an empty string ('').

`2. .astype(str)`
- Ensures all values in 'Phone_Number' are converted to strings.

`3. .str.replace(r'[^0-9]', '', regex=True)`
- Uses regular expressions (regex=True) to replace all non-numeric characters ([^0-9]) with an empty string ('').

```python
for x in df.index:
    if df.loc[x, "Do_Not_Contact"] == 'Y':
        df.drop(x, inplace=True)
```

    How It Works:
- Loops through each row index (x) in df.index.
- Checks if the value in the "Do_Not_Contact" column is 'Y'.
- If 'Y', it drops the row using df.drop(x, inplace=True).
