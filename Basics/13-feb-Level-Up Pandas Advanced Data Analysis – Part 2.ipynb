# 🔹 1. Efficient Data Loading & Memory Optimization

```python
df = pd.read_csv(r"I:\My Drive\Nareshit DS\Regular Notes\0-feb\13-feb\data.csv",low_memory=False, dtype={'CountryCode': 'category'})
```

```python
df.memory_usage(deep=True).sum() / (1024**2)  # Convert bytes to MB
```

# 🔹 2. Better Missing Data Handling

```python
df.isna().mean() *100 #Signature: df.isna() -> 'DataFrame'Docstring:Detect missing values.
# shows % of value per column
```

```python
df.dropna(axis=0, thresh=3) # drops rows with less than 3 valid valuse
```

1. .dropna()  # Removes rows or columns with missing (NaN) values
2. axis = 0 # Drops rowas
3. axis = 1 #(use axis=1 to drop column)
4. thresh = 3 # keeps rows that have at least 3 non-NaN values .
   - ** if a row has less than 3 valid values , it will be remove

```python
df.fillna(df.median(numeric_only=True), inplace=True) # fills with median Value 
```

# 🔹 3. Using .query() for Faster Filtering

```python
df[df['BirthRate']> 30]
```

1. df['BirthRate'] > 30 creates a boolean mask (True/False values).
2. "BirthRate > 30" is evaluated as a string.
3. Pandas processes it internally using a faster engine, improving performance.

```python
df.query("BirthRate > 30")
```
