```python
# Print header
print(" A     B   |   A AND B   |   A OR B   |   NOT A")
print("---------------------------------------------")

# Possible boolean values
values = [True, False]

# Generate truth table
for A in values:
    for B in values:
        and_result = A and B
        or_result = A or B
        not_result = not A  # NOT operation only on A
        
        print(f" {A:<5} {B:<5} |   {and_result:<9} |   {or_result:<8} |   {not_result:<5}")
```
