### Selecting the 'date' Column

rd['date']

This selects the **'date'** column from the DataFrame `rd`, which contains date strings.

### Applying Regular Expression (Regex)
```python
.str.extract(r'(\d+)(st|nd|rd|th)?\s(\w+)\s+(\d+)', expand=True)
```
This applies a **regular expression (regex)** to extract parts of the date string.

### Regex Pattern Breakdown
- `(\d+)` → Captures **one or more digits** (day part).
- `(st|nd|rd|th)?` → Optionally captures **ordinal suffixes** like **"st", "nd", "rd", or "th"**.
- `\s` → Matches a **space**.
- `(\w+)` → Captures the **month name** (a sequence of word characters).
- `\s+` → Matches **one or more spaces**.
- `(\d+)` → Captures **one or more digits** (year part).

### `expand=True`
Ensures that the extracted components are returned as a **new DataFrame** with multiple columns.


````python
cln_date = rd['date'].str.extract(r'(\d+)(st|nd|rd|th)?\s(\w+)\s+(\d+)', expand=True)
cln_date
````

````python
cln_date = cln_date.iloc[:,[0,2,3]]
cln_date
````

````python
cln_date.columns = ["Date", "Month", "Year"]
cln_date
````

````python
rd["time"] = rd["time"].astype(str)
cln_time = rd["time"].str.split(r'[:\s]+',expand = True)
cln_time
````

````python
cln_time.columns = ["Hrs","Mins","sec",""]
cln_time
````
