### 1. **`%matplotlib inline`**
   - This is a **magic command** used in **Jupyter Notebooks**.
   - It ensures that **plots are displayed directly inside the notebook** instead of opening in a separate window.

### 2. **`plt.rcParams['figure.figsize'] = 8, 4`**
   - `plt.rcParams` is a **dictionary-like object** in Matplotlib that allows you to configure global settings.
   - `'figure.figsize'` controls the **default size of figures** in inches (width, height).
   - Here, `8, 4` means the figure will be **8 inches wide and 4 inches tall**.
