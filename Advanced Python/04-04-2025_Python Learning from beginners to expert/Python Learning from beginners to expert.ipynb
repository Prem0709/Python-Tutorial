## 🔹 Variable-length Arguments

**variable-length arguments** in Python — `*args` and `**kwargs`. These are super useful when you’re not sure how many arguments will be passed to a function.


## 🔹 What are `*args` and `**kwargs`?
When defining functions:

`*args` lets you pass a variable number of positional arguments (they become a tuple).

`**kwargs` lets you pass a variable number of keyword arguments (they become a dictionary).



## 🔸 1. `*args` (Non-keyworded Variable Arguments)
✅ When to use:
- When you want your function to accept any number of positional arguments.

ex.

````python
def show_arg (*args):
    print(args)
    return args

reg = show_arg(11,12,13,14,15)
reg
````

````python
def add_args(*arg):
    total = 0
    for num in arg:
        total += num
    return total

first2 = add_args(1, 2)         # 3
print("Sum of first2:", first2)

all_sum = add_args(1, 2, 3, 4, 5)  # 15
print("Sum of all numbers:", all_sum)
````

````python
def add_arg(*arg):
    if len(arg) < 3:
        print ("pass more than 3 values")
        return
    
    a, b, c = arg[:3]  # Unpack first 3 values
    print(f'A = {a}')
    print(f'B = {b}')
    print(f'C = {c}')

add_arg(12,13,14)
````

# Global variable vs Local Variable


### 🔸 What is a Variable?
- A variable is a name that refers to a value stored in memory.

- But where that variable is defined decides its scope, which means where you can access it from in your code.



## 🔹 Local Variable
- `📌 Definition:` A local variable is defined inside a function and is only accessible within that function.

### exapmle:

````python
def my_function():
    x = 10  # Local variable
    print("Inside function:", x)

my_function()
print("Outside function:", x)  # ❌ This will raise an error!
````

````python
def add_value(a,b):
    c= a # C is Local variable
    d= b # d is Local variable
    sum = c+d
    return sum
add_value(12,21)
````

## 🔹 Global Variable
- `📌 Definition:`
A global variable is defined outside all functions and is accessible throughout the file, including inside functions (unless shadowed).
#### ✅ Example:

````python
x = 50  # Global variable

def my_function():
    c = x + x
    print("Inside function:", c)

my_function()
print("Outside function:", x)
````

### 🔹 **Anonymous Function** (also called **Lambda Function**)

> A **function without a name** is called an **anonymous function**.  
> In Python, we define it using the `lambda` keyword.

---

- Often used with functions like `map()`, `filter()`, and `sorted()`.
- Saves lines of code for simple operations.

---

### 🧾 Syntax of Lambda Function:
```python
lambda arguments : expression
```


````python
add = lambda x, y : x + y
print("Addition of a+ b =",add(5,6))
````

````python
square = lambda a : a*a
print(f"Square of A is ={square(6)}")
````


How can we use lambda in other function like - filter, map & reduce



### 1️⃣ Lambda with No Variables Assigned (True Anonymous)

````python
print ((lambda a,b: a+b)(23,2))
````

Here the lambda function is used only once — without assigning it to a variable.

### 2️⃣ Lambda with filter() – Filtering Data

````python
num = [1,2,3,4,5,6,78,9,0]
even = list(filter(lambda x: x%2 == 0, num))
print(even)
````

✅ filter() keeps only those items where the lambda returns True.

````python
odd= list(filter(lambda y: y%2 != 0, num))
print(odd)
````

````python
fi = 1,23,3,45,6,7, 8
type(fi)
even = tuple(filter(lambda x: x %2 ==0,fi))
print(even)
````

### 3️⃣ Lambda with map() – Transforming Data

````python
num
````

````python
add = list(map(lambda x : x+5, num))
print(add)
````

✅ map() applies the lambda function to each item in the list.
