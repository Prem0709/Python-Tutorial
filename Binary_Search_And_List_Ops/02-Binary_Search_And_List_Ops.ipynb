### 1️⃣ Binary Search
👉 Definition:
Binary search is an algorithm to find an element in a sorted list by repeatedly dividing the search interval in half.

Start with low = 0 and high = len(lst)-1

Find middle index → mid = (low + high) // 2

Compare lst[mid] with key:

If equal → found ✅

If key < lst[mid] → search in left half

If key > lst[mid] → search in right half

⚡ Condition: List must be sorted

```python
lst = [3,4,5,6,7,8]
low, high = 0, len(lst)-1
key = 6
# Iterative binary search
def binary_search(lst, key):
    low, high = 0, len(lst)-1
    while low <= high:
        mid = (low+high)//2
        if lst[mid] == key:
            return True
        elif key > lst[mid]:
            low = mid + 1
        else:
            high = mid - 1
    return False

print(binary_search(lst, 6))  # True

# 🔑 Time Complexity: O(log n)
# 🔑 Space Complexity: O(1)
```

### 2️⃣ List Insert (Custom Implementation)
- 👉 Normally, list.insert(index, value) inserts value at given index.


```python
def my_insert(lst, index, value):
    result = []
    for i in range(0, index):   # copy till index
        result.append(lst[i])
    result.append(value)        # insert element
    for j in range(index, len(lst)):  # append rest
        result.append(lst[j])
    return result

lst = [3, 4, 5, 67, 6, 7, 8, 9]
print(my_insert(lst, 4, 1000))
```

### 3️⃣ Count Occurrences (Recursive)

- 👉 Count how many times an element appears in a list.

```python
def recursive_my_count(lst, element):
    if lst == []:
        return 0
    elif lst[0] == element:
        return 1 + recursive_my_count(lst[1:], element)
    else:
        return recursive_my_count(lst[1:], element)

lst = [2,2,3,4,5,6,7,7,9,15]
print(recursive_my_count(lst, 7))   # 2
# 🔑 Time Complexity: O(n)
```

## 4️⃣ Binary Search (Recursive)
- 👉 Same logic as iterative but using recursion.

```python
def binary_search_recursive(lst, low, high, key):
    if low > high:
        return False
    mid = (low+high)//2
    if lst[mid] == key:
        return True
    elif lst[mid] > key:
        return binary_search_recursive(lst, low, mid-1, key)
    else:
        return binary_search_recursive(lst, mid+1, high, key)

lst = [3,4,5,6,7,8,9]
print(binary_search_recursive(lst, 0, len(lst)-1, 6))  # True
# 🔑 Time Complexity: O(log n)
# 🔑 Space Complexity: O(log n) (recursive stack).
```

### 5️⃣ Greatest from Left

- 👉 For each element, find the greatest element on its left.

- Q: For each element in the list, find the greatest element on its left side. If no greater element exists, put -1.

```python
def greatest_from_left(lst):
    result = []
    stack = []   # will store potential "greater" elements
    
    for num in lst:
        # remove elements smaller or equal to current
        while stack and stack[-1] <= num:
            stack.pop()
        
        if not stack:
            result.append(-1)
        else:
            result.append(stack[-1])
        
        stack.append(num)   # push current element for next iterations
    
    return result

lst=[70,30,50,45,50,46,78,67,70,75,100]
print(greatest_from_left(lst))
```

### 6️⃣ List Copy vs Slice
- 👉 Slicing makes a shallow copy but not linked with original.

```python
lst = [3,4,5,6,7,8,9,10]
lst1 = lst[1:6]   # slicing copy

print(id(lst), id(lst1))  # different IDs
lst1[4] = 999
print(lst, lst1)
```

#### 📒 Shallow Copy vs Deep Copy (Python)
##### 1. Shallow Copy

- 👉 Definition:
- A shallow copy creates a new object, but does not recursively copy inner/nested objects.

```python
import copy
lst1 = [[1, 2], [3, 4]]
shallow = copy.copy(lst1)
shallow[0][0] = 999
print("Original:", lst1)   # [[999, 2], [3, 4]]
print("Shallow:", shallow) # [[999, 2], [3, 4]]
```

##### 2. Deep Copy
- 👉 Definition:
- A deep copy creates a completely new object and recursively copies all inner/nested objects.

```python
import copy
lst1 = [[1, 2], [3, 4]]
deep = copy.deepcopy(lst1)
deep[0][0] = 999
print("Original:", lst1)  # [[1, 2], [3, 4]]
print("Deep:", deep)      # [[999, 2], [3, 4]]
```

### 7️⃣ Regex Example – Masking Numbers
- 👉 Replace digits with # except first and last.

```python
import warnings
warnings.filterwarnings("ignore")
import re
string="8888-7777-6666"
pattern="(\d)(\d+)(\-)(\d+)(\-)(\d{3})(\d)"
match=re.search(pattern,string)

masked = (
    match.group(1) + len(match.group(2))*"#" +
    match.group(3) + len(match.group(4))*"#" +
    match.group(5) + len(match.group(6))*"#" +
    match.group(7)
)
print(masked)  # 8###-####-###6
# 🔑 Time Complexity: O(n)
```

### 8️⃣ Valid Email using Regex
- 👉 Check if email format is correct.

```python
import re
def valid_email(email):
    pattern = "[a-zA-Z]+[0-9\_\.a-z]+\@[a-z]+\.[a-z]+"
    if re.findall(pattern, email):
        return True
    else:
        return False

print(valid_email("mayur.kulkarni636@gmail.ac.in"))  # True
```
