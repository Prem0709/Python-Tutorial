# String & Substring Notes (with Code Examples)

1. String Immutability in Python
   - A string is a sequence of characters enclosed in quotes (" " or ' ')
   - Example: "Prem", 'abc123'
   - Strings in Python are immutable (cannot be changed in-place).
   - p = "Prem"
   - p[1] = "p" ❌  --> ERROR (not allowed)

```python
 p = "Prem"
 p[1] = "p" ❌  --> ERROR (not allowed)
```

```python
s = "Prem"
s = s[:1] + "p" + s[2:]   # "Ppem"
s    
# 🔑 Note: Whenever you modify a string → a new string object is created → old one stays in memory until garbage collected.
# Time Complexity: Creating new string = O(n)
```

## 👉 Joining two strings together using +

```python
s1 = "Prem"
s2 = "Kumar"
print(s1 + s2)   # PremKumar
# ✔ Concatenation = O(n+m) (new string created).
```

### String Identity (is vs ==)

```python
s2 = "abc"
s3 = "abc"

print(s2 is s3)   # True (string interning for small strings)
print(s2 == s3)   # True (value equality)
print(id(s2), id(s3))  # Same memory address
# 🔑 Note:
# is → compares memory address
# == → compares values
```

```python
# 👉 Generate all substrings using two loops:
```

```python
def all_substrings(string):
    for i in range(len(string)):
        for j in range(i+1, len(string) + 1):
            print(string[i:j])
s = "abc"
all_substrings(s)
## Time Complexity: O(n³) (nested loops + slicing).
```

🔹 Optimized Approach (O(n²))

```python
def all_substrings(string):
    n = len(string)
    for i in range(n):
        sub = ""
        for j in range(i, n):
            sub += string[j]   # build incrementally
            print(sub)

s = "abc"
all_substrings(s)
```

### All Palindromic Substrings

```python
# 👉 A palindrome = same forward & backward.
```

```python
def ispallindrome(s):
    i, j = 0, len(s)-1
    while i < j:
        if s[i] != s[j]:
            return False
        i, j = i+1, j-1
    return True
def all_pallindromic_substrings(string):
    for i in range(len(string)):
        for j in range(i+1, len(string)+1): 
            ss = string[i:j]
            if ispallindrome(ss):
                print(ss)
s = "ababa"
all_pallindromic_substrings(s)
```

```python
def all_palindromic_substrings(s):
    n = len(s)
    def expand(l, r):
        while l >= 0 and r < n and s[l] == s[r]:
            print(s[l:r+1])   # palindrome found
            l -= 1
            r += 1
    for center in range(n):
        # Odd-length palindromes (center at one char)
        expand(center, center)
        # Even-length palindromes (center between two chars)
        expand(center, center+1)
s = "ababa"
all_palindromic_substrings(s)
```

## Definition: String Compression
- 👉 Reducing string size by removing or counting consecutive duplicate characters.

```python
# def compress_substring(s):
#     prev = 0
#     net = prev + 1
#     result = s[0]
#     while net < len(s):
#         if s[prev] != s[net]:
#             result += s[net]
#         prev, net = net, net + 1
#     print(result)
# s = "aaabbbbaaaacccc"
# compress_substring(s)   # abac

def compress_substring(s):
    result = s[0]   # start with first char
    for i in range(1, len(s)):
        if s[i] != s[i-1] and s[i] not in result: #avoid duplicates  # only add if different from previous
            result += s[i]
    print(result)

s = "aaabbbbaaaacccc"
compress_substring(s)   # abc
```

### Extended Compression (Run Length Encoding)

```python
def string_compression_ext(s):
    result = ""
    count = 1
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            result += s[i-1] + str(count)
            count = 1
    result += s[-1] + str(count)
    return result

s = "aaaabbbbcccrccc"
print(string_compression_ext(s))   # a4b4c6
```

## Regular Expression (Regex)

- A Regular Expression (Regex) is a sequence of characters that forms a search pattern. 
- It is mainly used for pattern matching, validation, searching, and replacing text in strings.

`Example:
[a-z] → lowercase letters
\d → digits
.* → match everything`

| Symbol  | Meaning                      | Example     | Matches                                     |     |            |
| ------- | ---------------------------- | ----------- | ------------------------------------------- | --- | ---------- |
| `.`     | Any character except newline | `a.c`       | `abc`, `a1c`, `a-c`                         |     |            |
| `^`     | Start of string              | `^Hello`    | Matches `Hello world` but not `Say Hello`   |     |            |
| `$`     | End of string                | `world$`    | Matches `Hello world` but not `world peace` |     |            |
| `*`     | 0 or more repetitions        | `a*`        | \`\`, `a`, `aa`, `aaa`                      |     |            |
| `+`     | 1 or more repetitions        | `a+`        | `a`, `aa`, `aaa` (but not empty)            |     |            |
| `?`     | 0 or 1 repetition            | `a?`        | \`\`, `a`                                   |     |            |
| `{n}`   | Exactly n repetitions        | `a{3}`      | `aaa`                                       |     |            |
| `{n,m}` | Between n and m repetitions  | `a{2,4}`    | `aa`, `aaa`, `aaaa`                         |     |            |
| `[]`    | Any one character inside     | `[abc]`     | `a`, `b`, `c`                               |     |            |
| `[^]`   | Negation inside brackets     | `[^abc]`    | Any character except a, b, c                |     |            |
| \`      | \`                           | OR operator | \`a                                         | b\` | `a` or `b` |
| `()`    | Grouping                     | `(abc)+`    | `abc`, `abcabc`                             |     |            |
| `\`     | Escape special char          | `\.`        | Matches `.` literally                       |     |            |


Quantifiers

`* → 0 or more`

`+ → 1 or more`

`? → 0 or 1`

`{n} → exactly n times`

`{n,m} → n to m times`

```python
import re
s = "abababa"
p = ".*"
r = re.findall(p, s)
r
```

## Regex Example: Email Extraction

```python
s = "prempawar0709@gmail.com"
p = r"([a-zA-Z0-9])"
# p = r"(\@[a-zA-Z0-9]+)"
# p= r"(\.)([a-zA-Z0-9]+)"
match = re.search(p, s)
match.group()
```

```python
s = "prempawar0709@gmail.com"
p = r"([a-zA-Z0-9]+)(\@[a-zA-Z0-9]+)(\.)([a-zA-Z0-9]+)"
match = re.search(p, s)

print(match.group(1))  # prempawar0709
print(match.group(2))  # @gmail
print(match.group(3))  # .
print(match.group(4))  # com
```

```python
s = "prempawar0709@gmail.com"
p = r"([a-z])([a-zA-Z0-9]+)(\@)([a-z]+)(\.)([a-zA-Z0-9]+)"
match = re.search(p, s)

result = match.group(1) + len(match.group(2))*"#"
masked_email = result + match.group(3)+match.group(4)+match.group(5)+match.group(6)

print(masked_email)  # m###############@gmail.com
```
